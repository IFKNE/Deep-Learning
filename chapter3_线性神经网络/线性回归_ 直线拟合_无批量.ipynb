```python
import numpy as np #导入必要的运行库
import pandas as pd

#np.random.seed(42)      #设置随机数种子，保证每次运行结果一致（可选）

a  = np.random.randn(2)

df = pd.read_csv('Salary_Data.csv') #你可以通过pandas读取本文件夹内的数据，也可以自行构建数据或是使用kaggle的数据集，教程在实践引导中
learning_rate = 0.1 #学习率
epochs = 2000 #迭代次数

if_normalize=1 #如果x和y都进行标准化，则设置if_normalize=1
if if_normalize==1: #标准化，即 (x-均值)/标准差，将x，y化为0均值，1方差分布
    df_unnormalized = df.copy()
    df['YearsExperience'] = (df['YearsExperience']-df['YearsExperience'].mean())/df['YearsExperience'].std()
    df['Salary'] = (df['Salary']-df['Salary'].mean())/df['Salary'].std()

for i in range(epochs):
    loss_vector = df['Salary']-a[0]*df['YearsExperience']-a[1]
    loss = np.sum(loss_vector**2)/len(loss_vector)  #损失函数


    gradient_a0 = -2*np.mean((df['Salary']-a[0]*df['YearsExperience']-a[1])*df['YearsExperience']) #用数学方法计算梯度
    gradient_a1 = -2*np.mean(df['Salary']-a[0]*df['YearsExperience']-a[1])


    a[0] = a[0] - learning_rate*gradient_a0 #更新参数
    a[1] = a[1] - learning_rate*gradient_a1

    if i % 100 == 0: #每100个epoch打印一次损失
        print(f'Epoch {i}, Loss: {loss}')

if if_normalize==1: #如果x和y都进行标准化，则需要将参数a进行还原
    a[0] = a[0] * df_unnormalized['YearsExperience'].std() + df_unnormalized['YearsExperience'].mean()
    a[1] = a[1] * df_unnormalized['Salary'].std() + df_unnormalized['Salary'].mean()

print(a)
print(loss)
```

```python
import torch
a  = torch.randn(2)
data = pd.read_csv("data.csv")
print(data)
```

```python
my_y=data.iloc[:,0]
my_x=data.iloc[:,1:3]
print(my_y)
print(my_x)
```

```python
data_x=torch.tensor(my_x.to_numpy(dtype=float))
data_y=torch.tensor(my_y.to_numpy(dtype=float))
data_x=data_x.T
print(data_x)
print(data_y)
```

```python
weight = torch.randn(2,dtype=float, requires_grad=True) #随机初始化权重，并设置requires_grad=True以便后续计算梯度
bias = torch.randn(1, dtype=float, requires_grad=True)  #随机初始化偏置，并设置
print(weight,weight.shape)
print(bias,bias.shape)
print(data_x,data_x.shape)
weight.T@data_x+bias
```

```python
def loss(weight, bias, x, y):
    return torch.mean((weight.T@x+bias-y)**2) #定义损失函数 
my_loss=loss(weight, bias, data_x, data_y)
print(my_loss)
```

```python
my_loss.backward() #反向传播计算梯度
print(weight.grad) #打印权重的梯度
print(bias.grad) #打印偏置的梯度
```

```python
lr=0.001 #学习率
times=100 #迭代次数
while   times>0:
    my_loss=loss(weight, bias, data_x, data_y) #计算损失
    my_loss.backward() #反向传播计算梯度
    with torch.no_grad(): #更新参数
        weight -= lr*weight.grad
        bias -= lr*bias.grad
    weight.grad.zero_() #清零梯度
    bias.grad.zero_()
    if times % 10 == 0: #每10个epoch打印一次损失
        print(f'Epoch {100-times:10.0f},w0: {weight[0].item():5.4f}, w1: {weight[1].item():5.4f}, b: {bias.item():5.4f}, Loss: {my_loss.item():5.4f}')
    times-=1
```
