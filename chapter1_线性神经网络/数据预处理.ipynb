```python
import torch
x=torch.arange(24).reshape(2,3,4)
print("x=")
print(x),print("x.shape=",x.shape)
print("x.sum()=")
print(x.sum()),print("x.sum().shape=",x.sum().shape)
print("x.sum(axis=0)=")
print(x.sum(axis=0)),print("x.sum(axis=0).shape=",x.sum(axis=0).shape)
print("x.sum(axis=1)=") 
print(x.sum(axis=1)),print("x.sum(axis=1).shape=",x.sum(axis=1).shape)
print("x.sum(axis=2)=")
print(x.sum(axis=2)),print("x.sum(axis=2).shape=",x.sum(axis=2).shape)
print("x.sum(axis=[0,1])=")
print(x.sum(axis=[0,1])),print("x.sum(axis=[0,1]).shape=",x.sum(axis=[0,1]).shape)
print("x.sum(axis=[0,2])=")
print(x.sum(axis=[0,2])),print("x.sum(axis=[0,2]).shape=",x.sum(axis=[0,2]).shape)
print("x.sum(axis=[1,2])=")
print(x.sum(axis=[1,2])),print("x.sum(axis=[1,2]).shape=",x.sum(axis=[1,2]).shape)
print("x.sum(axis=[0,1,2])=")
print(x.sum(axis=[0,1,2])),print("x.sum(axis=[0,1,2]).shape=",x.sum(axis=[0,1,2]).shape)
print("x.sum(axis=0, keepdims=True)=")
print(x.sum(axis=0, keepdims=True)),print("x.sum(axis=0, keepdims=True).shape=",x.sum(axis=0, keepdims=True).shape)
print("x.sum(axis=1, keepdims=True)=")
print(x.sum(axis=1, keepdims=True)),print("x.sum(axis=1, keepdims=True).shape=",x.sum(axis=1, keepdims=True).shape)
print("x.sum(axis=2, keepdims=True)=")
print(x.sum(axis=2, keepdims=True)),print("x.sum(axis=2, keepdims=True).shape=",x.sum(axis=2, keepdims=True).shape)
```
